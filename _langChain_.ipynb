```python
## 2. Imports & Setup
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig, pipeline
from langchain_community.llms import HuggingFacePipeline
```

```python
## 3. Load the Model
#We use the `chavinlo/alpaca-native` model in 4‑bit mode to save memory and run efficiently:

model_id = "chavinlo/alpaca-native"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="auto",
    load_in_4bit=True,
    torch_dtype="auto",
)
```

```python
## 4. Create a Hugging Face Pipeline
#Wrap the model into a text-generation pipeline:

hf_pipeline = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_length=256,
)

llm = HuggingFacePipeline(pipeline=hf_pipeline)
```

```python
## 5. testing the model
prompt = "Write a short story about a ml and dl with comparason."
response = llm(prompt)

print("📜 Response:\n", response)
```

```python
## 5. Integrate with LangChain
# You can create a prompt template by defining a string with placeholders
#  and using PromptTemplate.from_template().

from langchain.prompts import PromptTemplate

template = "Write a short story about a {animal} who loves {activity}."
prompt = PromptTemplate.from_template(template)
```

```python
#To generate the final prompt string, use the format() method with the appropriate values
# for the placeholders

formatted_prompt = prompt.format(animal="cat", activity="painting")
print(formatted_prompt)
```

```python
#Once you have the formatted prompt, you can pass it to an LLM for processing

response = llm(formatted_prompt)
print(response)
```

# **Example usage Langchain**
LangChain supports few-shot learning by allowing you to include examples in your prompt templates. This can help the model understand the desired output format or style

```python
# take example of translate to see output and compare after Implementing Few-Shot Prompting
response = llm("Translate 'Hello' to French.")
print(response)
```

**Few-Shot Prompt Templates**

is a technique where you supply the model with a few examples of the desired behavior. This helps the model understand the task better, especially when it's not performing as expected with zero-shot prompts. In LangChain, the FewShotPromptTemplate class facilitates this by allowing you to define:​
Pinecone

Examples: A list of dictionaries containing input-output pairs.

Example Prompt: A PromptTemplate that formats each example.

Prefix/Suffix: Strings that wrap around the examples, providing context or instructions.

```python
from langchain.prompts import FewShotPromptTemplate

# 1. Define the Examples
examples = [
    {"input": "Translate 'Thank you' to French.", "output": "merci"},
    {"input": "Translate 'Goodbye' to French.", "output": "Au revoir"}
]

# 2. Create the Example Prompt Template
example_prompt = PromptTemplate.from_template("Input: {input}\nOutput: {output}")

# 3. Construct the Few-Shot Prompt Template
few_shot_prompt = FewShotPromptTemplate(
    examples=examples,  # The list of example translations
    example_prompt=example_prompt,  # How each example is formatted
    prefix="Translate the following sentences to French:",  # Instruction before examples
    suffix="Input: Translate '{sentence}' to French.\nOutput:",  # Prompt for the new input
    input_variables=["sentence"]  # Variable to be filled with the new sentence
)

# 4. Format the Prompt with New Input
formatted_prompt = few_shot_prompt.format(sentence="Good morning")

response = llm(formatted_prompt)
print(response)
```
